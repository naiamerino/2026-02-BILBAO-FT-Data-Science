Máquina virtual. Una capa 'intermedia' a la que le adjudicamos parte de los recursos del sistema y sobre la que ejecutamos ciertas instrucciones. Como hacer particiones de ordenador  
Docker ---> hemos instalado un Windows Subsystem Layer? que está ejecutando linux y sobre el que corre Docker 

Docker es un so para ejecutar contenedores. ¿Qué son contenedores? 'Particiones' más pequeñas que una máquina virtual en la que corren aplicaciones o partes de aplicaciones.

Pasos:
1- Creamos un dockerfile: es un archivo de texto que contiene 'las instrucciones' .  
2- Hago un docker build sobre ese archivo y genero la imagen  
3- Ejecutando un docker run sobre esa imagen crearemos ese contenedor
